## Tunisia

Only the .shp, .shx, .dbf, .prj are necessary for your analysis
| Extension | Description                                                              |
| --------- | ------------------------------------------------------------------------ |
| `.shp`    | The main geometry file (points, lines, polygons).                        |
| `.shx`    | The shape index (links geometry to attributes).                          |
| `.dbf`    | The attribute table (like a spreadsheet with land cover info).           |
| `.prj`    | The projection info (CRS — coordinate system).                           |
| `.avl`    | A styling/symbology file (for ArcGIS/MapInfo — not needed for analysis). |
| `.lyr`    | Another layer style file (used by ArcGIS — can ignore).                  |


```python
import geopandas as gpd

# Path to the .shp file
tn_path = f"{landCover_folder}/Tunisia/geonetwork_landcover_tun_gc_adg/tun_gc_adg.shp"

# Load it
lc_tn = gpd.read_file(tn_path)

print(lc_tn.head())
print(lc_tn.crs)
```

You’re essentially extracting or converting the Tunisia land cover dataset from its raw shapefile format to a clean, ready-to-use GeoJSON — which can then be:

- Clipped to another geometry (like administrative boundaries),

- Used for visualization (maps, dashboards),

- Combined with raster data (for land cover classification).

```python
import geopandas as gpd

# Paths to your GeoJSON files
dz_path = f"{extracted_landcover_folder}/landcover_algeria.geojson"
tn_path = f"{extracted_landcover_folder}/landcover_tunisia.geojson"

# Read them
gdf_dz = gpd.read_file(dz_path)
gdf_tn = gpd.read_file(tn_path)

# --- 2. Standardize column names ---
# lowercase all column names to avoid case issues
gdf_dz.columns = gdf_dz.columns.str.lower()
gdf_tn.columns = gdf_tn.columns.str.lower()

# --- 3. Harmonize columns (if one has AREA and the other AREA_M2) ---
if "area" not in gdf_tn.columns and "area_m2" in gdf_tn.columns:
    gdf_tn = gdf_tn.rename(columns={"area_m2": "area"})

# --- 4. Match CRS ---
if gdf_dz.crs != gdf_tn.crs:
    gdf_tn = gdf_tn.to_crs(gdf_dz.crs)

# --- 5. Merge both datasets ---
merged = gpd.GeoDataFrame(
    pd.concat([gdf_dz, gdf_tn], ignore_index=True),
    crs=gdf_dz.crs
)

# --- 6. Optionally dissolve to get one merged boundary ---
merged_geom = merged.dissolve()  # merges into one geometry

# --- 7. Save results ---
merged.to_file(f"{extracted_landcover_folder}/landcover_combined.geojson", driver="GeoJSON")
merged_geom.to_file(f"{extracted_landcover_folder}/landcover_combined_boundary.geojson", driver="GeoJSON")

print("✅ Land cover data merged and saved successfully!")
```
