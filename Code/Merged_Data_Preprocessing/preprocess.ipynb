# reading the geojson file

```python
# read df_final_merged.geojson file in MergedDatasets
import geopandas as gpd
df_final_merged = gpd.read_file('../../MergedDatasets/df_final_merged.geojson')
```

# Filtering to keep mostly north 
--> altitude and longitude were studied and chosen in Data_Preprocesssing/fire.ipynb

```python
# Define the coordinate limits
lon_min = -2.5
lon_max = 11.11035
lat_min = 26
lat_max = 37.32346

# Get the bounding box coordinates for each geometry
bounds = df_final_merged.geometry.bounds

# Create a boolean mask:
# A row is kept if its minimum longitude (minx) is less than the max limit
# AND its maximum longitude (maxx) is greater than the min limit
# AND its minimum latitude (miny) is less than the max limit
# AND its maximum latitude (maxy) is greater than the min limit
# This ensures that any part of the polygon intersecting the box is kept.

mask = (bounds['minx'] < lon_max) & \
       (bounds['maxx'] > lon_min) & \
       (bounds['miny'] < lat_max) & \
       (bounds['maxy'] > lat_min)

# Apply the mask to filter the GeoDataFrame
df_filtered = df_final_merged[mask].copy()

print(f"Original number of rows: {len(df_final_merged)}")
print(f"Filtered number of rows: {len(df_filtered)}")
print("✅ GeoDataFrame successfully filtered by spatial bounds.")
print("Percentage of data retained: ")
percentage_retained = (len(df_filtered) / len(df_final_merged)) * 100
print(f"{percentage_retained:.2f}%")
```

```python
# viewing number of rows where fire=1 and fire=0
print(df_filtered['fire'].value_counts())

# percentage of rows where fire=1
fire_percentage = (df_filtered['fire'].sum() / len(df_filtered)) * 100
print(f"Percentage of rows with fire=1: {fire_percentage:.2f}%")
print(f"Percentage of rows with fire=0: {100 - fire_percentage:.2f}%")
```
